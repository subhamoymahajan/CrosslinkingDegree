# S1. Genral Polymer Equations

Eq. S1. $n_{M_1} + n_{M_2} + n_{T_1} + n_{T_2} + n_{T_3} = 1$

Eq. S2. $n_{M_1} +  2n_{M_2} = n_{T_1} + 2n_{T_2} + 3n_{T_3}$

```python
from sympy import *
```

```python
v_M1 = symbols('n_{M_1}')
v_M2 = symbols('n_{M_2}')
v_T1 = symbols('n_{T_1}')
v_T2 = symbols('n_{T_2}')
v_T3 = symbols('n_{T_3}')
```

```python
#Eq. S1 = 0
Eq_S1 = v_M1 + v_M2 + v_T1 + v_T2 + v_T3 -1
#Eq. S2 = 0
Eq_S2 = v_M1 + 2*v_M2 - v_T1 -2*v_T2 -3*v_T3
```

Eq. S3:    $ \theta = \frac{ n_{T_1}}{ n_{T_1} + n_{M_1}}$ 

Eq. S4:    $ n = \frac{ n_{T_3}}{ n_{T_3} + n_{T_2}}$ 

```python
#th = T1/(T1+M1)
v_th = symbols('theta')
Eq_S3 = v_th*(v_T1+v_M1)-v_T1

#n = T3/(T2+T3)
v_n = symbols('n')
Eq_S4 = v_n*(v_T2+v_T3)-v_T3

#e = (T3-T1-M1)/T3
v_e = symbols('epsilon')
Eq_S6 = v_e*v_T3-v_T3+v_T1+v_M1
```

```python
#Solving equations for n, e, theta
sol = solve([Eq_S1, Eq_S2, Eq_S3, Eq_S4, Eq_S6], (v_M1, v_M2, v_T1, v_T2, v_T3))
# M1
M1_1 = collect(simplify(collect(collect(sol[v_M1],v_n),v_th)),1-v_e)
M1_1_num = numer(M1_1)
M1_1_den = denom(M1_1)
M1_1_den = expand(-M1_1_den) # -negative
M1_1_den = collect(collect(collect(M1_1_den,v_n),v_e),2*v_th+1)
M1_1 = (-M1_1_num)/M1_1_den
eqn_m1 = Eq(v_M1, M1_1)
display(eqn_m1)

# M2 
M2_1 = collect(collect(collect(collect(sol[v_M2],v_n),v_e),2*v_th+1),2*v_th-1)
eqn_m2 = Eq(v_M2, M2_1)
display(eqn_m2)

#T1
T1_1 = simplify(collect(collect(collect(collect(sol[v_T1],v_n),v_e),2*v_th+1),v_th))
T1_1_num = simplify(-numer(T1_1))
T1_1_den = expand(-denom(T1_1))
T1_1_den = collect(collect(collect(T1_1_den,v_n),v_e),2*v_th+1)
T1_1 = T1_1_num/T1_1_den
eqn_t1 = Eq(v_T1, T1_1)
display( eqn_t1)

#T2 
T2_1 = collect(collect(collect(sol[v_T2],v_n),v_e),2*v_th+1)
T2_1_num = collect_const(simplify(-numer(T2_1)))
T2_1_den = expand(-denom(T2_1))
T2_1_den = collect(collect(collect(T2_1_den,v_n),v_e),2*v_th+1)
T2_1 = T2_1_num/T2_1_den
eqn_t2 = Eq(v_T2,T2_1)
display(eqn_t2)

#T3 
T3_1_num = -numer(sol[v_T3])
T3_1_den = -denom(sol[v_T3])
T3_1_den = collect(collect(collect(T3_1_den,v_n),v_e),2*v_th+1)
T3_1 = T3_1_num/T3_1_den
eqn_t3 = Eq(v_T3, T3_1)
display(eqn_t3)
```

# S2. General Polymer Equation

```python
#Eq S9
v_ron = symbols('r_{O/N}')
rON = (5*T1_1 + 4*T2_1 + 3*T3_1)/(2*M1_1 + 2*M2_1)
rON_1 = collect(collect(simplify(rON),v_n),v_th)
rON_1_num = expand(-numer(rON_1))
rON_1_den = expand(-denom(rON_1))
rON_1_num = collect(collect(rON_1_num,v_n),5*v_th)
rON_1_den = collect(rON_1_den,v_n)
rON_1 = rON_1_num / rON_1_den
eqn_ron = Eq(v_ron, rON_1)
display(eqn_ron)
```

Eq. S9: $r_{O/N} = \frac{4-n+5n\theta(1-\epsilon)}{2+n(2-\epsilon)}$

```python
Eq_S9 = v_ron*(2*v_M1+2*v_M2) - 5*v_T1 -4*v_T2 -3*v_T3
```

```python
#Solving equations for n, e, theta
sol = solve([Eq_S1, Eq_S2, Eq_S9], (v_M1, v_M2, v_T1))

#M1 
M1_2_num = expand(numer(sol[v_M1]))
q = Dummy('q')
target = v_T2 + 2*v_T3 + 1
# Rewrite T2 using q = T2 + 2*T3 + 1
expr = expand(M1_2_num.subs(v_T2, q - 2*v_T3 - 1))
# Collect powers of q, then restore the original expression
M1_2_num = collect(expr, q, func=factor).subs(q, target)
M1_2_den = denom(sol[v_M1])
M1_2 = M1_2_num/M1_2_den
eqn_m1 = Eq(v_M1, M1_2)

display(eqn_m1)

#M2
M2_2_num = expand(numer(sol[v_M2]))
q = Dummy('q')
target = v_T2 + 2*v_T3 + 1
# Rewrite T2 using q = T2 + 2*T3 + 1
expr = expand(M2_2_num.subs(v_T2, q - 2*v_T3 - 1))
# Collect powers of q, then restore the original expression
M2_2_num = collect(expr, q, func=factor).subs(q, target)
M2_2_den = denom(sol[v_M2])
M2_2 = M2_2_num/M2_2_den
eqn_m2 = Eq(v_M2, M2_2)
display(eqn_m2)

#T1
T1_2_num = factor_terms(collect(collect(expand(numer(sol[v_T1])),v_T2),v_T3))
T1_2_den = denom(sol[v_T1])
T1_2 = T1_2_num / T1_2_den
eqn_t1 = Eq(v_T1, T1_2)
display(eqn_t1)

#DC3
v_DC3 = symbols('DC_3')
DC3_num = collect(simplify(v_T3-(T1_2+M1_2)),v_ron)
#display(DC3_num)
q = Dummy('q')
target = v_T2 + 2*v_T3 
expr = simplify(collect(expand(DC3_num.subs(v_T2, q - 2*v_T3 )),q))
p = Dummy('p')
target2 = v_ron + 3
expr2 = collect(expr.subs(v_ron, p -3),p,func=factor)

DC3_num = expr2.subs(p,target2).subs(q,target)
DC3 = DC3_num/(v_T2+v_T3)
eqn_dc3 = Eq(v_DC3, DC3)
display(eqn_dc3)
```

# S3. General Polymer Equations with O/N and O$_2$/O characterization

```python
#Eq S12d
v_ro2 = symbols('r_{O_2}')
rO2 = (T1_1 + 2*T2_1 + 3*T3_1)/(5*T1_1 + 4*T2_1 +3*T3_1)
rO2_1 = simplify(rO2)
rO2_1_num = factor_terms(collect(expand(-numer(rO2_1)),v_th))
rO2_1_den = factor_terms(collect(expand(-denom(rO2_1)),v_th))

rO2_1 = rO2_1_num / rO2_1_den
eqn_ro2 = Eq(v_ro2, rO2_1)
display(eqn_ro2)
```

```python
Eq_S12d = v_ro2*(5*v_T1 + 4*v_T2 + 3*v_T3) - v_T1 -2*v_T2 -3*v_T3
#Solving equations for n, e, theta
sol = solve([Eq_S1, Eq_S2, Eq_S9, Eq_S12d], (v_M1, v_M2, v_T2, v_T3))

#M1 
M1_3 = collect(factor_terms(sol[v_M1]),v_ron)
eqn_m1 = Eq(v_M1, M1_3)
#display(eqn_m1)

#M2 
M2_3 = collect(factor_terms(sol[v_M2]),v_ron)
eqn_m2 = Eq(v_M2, M2_3)
#display(eqn_m2)

#T2 
T2_3 = collect(apart(sol[v_T2],v_T1),v_ron)

foo = Add.make_args(T2_3)
foo1_num = factor_terms(numer(foo[1]))
foo1_1 = list(Mul.make_args(foo1_num))

foo1_1[0] = -foo1_1[0]
foo1_1[2] = -foo1_1[2]
foo1_num  = foo1_1[0]*foo1_1[1]*foo1_1[2]

foo1_den = denom(foo[1])
T2_3 = foo[0] + foo1_num/foo1_den 
eqn_t2 = Eq(v_T2, T2_3)
#display(eqn_t2)

#T3 
T3_3 = factor_terms(collect(apart(sol[v_T3],v_T1),v_ron))
eqn_t3 = Eq(v_T3, T3_3)
#display(eqn_t3)

#DC3 
DC3_3 = collect(collect(simplify((T3_3 - v_T1 - M1_3)/(T3_3 + T2_3)),v_T1),v_ron)
DC3_3 = factor_terms(DC3_3)
display(DC3_3)
```

```python
# alpha/beta Eq. S15d
a_ = symbols('alpha')
b_ = symbols('beta')
r_ab = a_/b_
a_b = simplify((2*v_T1 + T2_3)/(M1_3+2*M2_3))
eqn_ab = Eq(r_ab,a_b)
display(eqn_ab)

# gamma/beta Eq. S15e
g_ = symbols('gamma')
r_gb = g_/b_
g_b = simplify( M1_3/(M1_3+2*M2_3))
eqn_gb = Eq(r_gb, g_b)
display(eqn_gb)
```

# S4. General Polymer Equations with FTIR and NMR characterization

```python
# r_NMR
v_rnmr = symbols('r_{NMR}')
rNMR = simplify((v_T1 + T2_3 +T3_3)/2)
eqn_nmr = Eq(v_rnmr, rNMR)
display(eqn_nmr)

# r_FTIR
v_rftir = symbols('r_{FTIR}')
rFTIR = simplify((2*v_T1 + T2_3)/(v_T1+2*T2_3 +3*T3_3))
eqn_ftir = Eq(v_rftir, rFTIR)
display(eqn_ftir)
```

```python
Eq_S17a = 2*v_rnmr*(3+v_ron*(v_ro2+1))-v_ron*(v_ro2+1)
Eq_S17b = v_rftir*2*v_ro2 -(1-v_ro2)


sol = solve([Eq_S17a, Eq_S17b], (v_ro2, v_ron))
rO2_4 = sol[0][0]
rON_4 = sol[0][1]
display( Eq(v_ro2, rO2_4))
display( Eq(v_ron, rON_4))
```

```python
#DC3 
DC3_4 = (6+2*v_ron*(1-5*v_ro2))/(v_T1*(3+v_ron*(v_ro2+1))-v_ron*(1+v_ro2))
DC3_4_1 = simplify(DC3_4.subs(v_ron,rON_4).subs(v_ro2,rO2_4))
display( simplify(collect(collect(DC3_4_1,v_rnmr),v_T1)))
```

# S5. General Polymer Equations with C/O/N proportions

```python
v_a=symbols('a')
v_b=symbols('b')
v_c=symbols('c')
v_d=symbols('d')

Eq_S19a = v_M1 - v_d
Eq_S19b = v_M2 - v_b + v_d
Eq_S19c = v_T1 + v_T2 + v_T3 - v_a
Eq_S19d = 2*v_T1 + v_T2 - v_c

sol = solve([Eq_S19a, Eq_S19b, Eq_S19c, Eq_S19d], (v_a, v_b, v_c, v_d))
display(Eq(a, sol[v_a]))
display(Eq(b, sol[v_b]))
display(Eq(c, sol[v_c]))
display(Eq(d, sol[v_d]))
a_5 = sol[v_a]
b_5 = sol[v_b]
c_5 = sol[v_c]
d_5 = sol[v_d]

v_rc = symbols('r_C')
v_ro = symbols('r_O')
v_rn = symbols('r_N')

# r_C = 9*a +6*b 
Eq_rc = v_rc - 9*a_5 -6*b_5
# r_O = 3*a + c
Eq_ro = v_ro - 3*a_5 -c_5
# r_N = 2b
Eq_rn = v_rn - 2*b_5
Eq_sum = v_rc + v_ro + v_rn - 1 
Eq_cons = 2*b_5 -3*a_5 + c_5 -d_5 
sol = solve([Eq_rc, Eq_ro, Eq_rn, Eq_sum, Eq_cons], (v_M1, v_M2, v_T1, v_T2))
#sol

M1_5 = sol[v_M1].subs(v_rc,1-v_ro-v_rn)
display(Eq(v_M1, M1_5))
M2_5 = sol[v_M2].subs(v_rc,1-v_ro-v_rn)
display(Eq(v_M2, M2_5))
T1_5 = sol[v_T1].subs(v_rc,1-v_ro-v_rn)
display(Eq(v_T1, T1_5))
T2_5 = sol[v_T2].subs(v_rc,1-v_ro-v_rn)
display(Eq(v_T2, T2_5))
DC3_5 = (v_T3-sol[v_M1]-sol[v_T1])/(v_T3+sol[v_T2])
DC3_5 = DC3_5.subs(v_rc, 1-v_ro-v_rn)
display(Eq(v_DC3,simplify(DC3_5)))

display('MPD/TMC', simplify((M1_5+M2_5)/(T1_5+T2_5+v_T3)))

rO2_5 =apart(simplify((T1_5+2*T2_5+3*v_T3)/(5*T1_5 + 4*T2_5 +3*v_T3) ), v_ro)
display(Eq(v_ro2, rO2_5))
```
